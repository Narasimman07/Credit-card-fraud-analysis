```python
df['Hour'] = (df['Time'] // 3600) % 24
```

```python
df
```

```python
df.groupby('Hour')['Class'].mean() * 100
```

```python
df['RiskLevel'] = df['Amount'].apply(
    lambda x: 'Low' if x < 50 else ('Medium' if x < 200 else 'High')
)
```

```python
df.groupby('RiskLevel')['Class'].mean() * 100
```

```python
from sklearn.ensemble import IsolationForest

model = IsolationForest(contamination=0.01, random_state=42)
df['anomaly'] = model.fit_predict(df[['Amount']])

df['anomaly'] = df['anomaly'].map({1:0, -1:1})
```

```python
from sklearn.metrics import classification_report
print(classification_report(df['Class'], df['anomaly']))
```

```python
sns.lineplot(x='Hour', y='Class', data=df)
plt.title("Fraud Trend by Hour")
plt.show()
```

```python
sns.boxplot(x='Class', y='Amount', data=df)
plt.show()
```
